# Вычисление логических функций средствами однослойных искусственных нейронных сетей

В качестве примера построим и обучим искусственную нейронную сеть (ИНС) вычислению конъюнкции.

Сначала подгрузим библиотеки с линейной моделью слоев Sequential, со слоями Dense

```python
import keras
from keras.models import Sequential
from keras.layers import Dense, Activation
from keras.optimizers import SGD
```

Создадим модель ИНС и добавим в нее один слой с одним нейроном (units = 1) и двумя входами (input_dim = 2)

```python
from keras import activations
model = Sequential()
model.add(Dense(units = 1, input_dim = 2, activation='sigmoid'))
```

Скомпилируем, получившуюся ИНС

```python
optimizer = keras.optimizers.SGD(learning_rate = 0.6)

model.compile(optimizer, loss='mse', metrics=['binary_accuracy'])
```

Создадим обучающий набор X и вектор значений целевой функции (конъюнкции)

```python
import numpy as np
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
X
```

```python
y = np.array([0, 0, 0, 1])
y
```

Обучим ИНС

```python
model.fit(X, y, epochs=5)
```

Протестируем ИНС

```python
model.predict(X)
```

1. Обучить ИНС для вычисления дизъюнкции

```python
# Генерируем обучающие данные для дизъюнкции

Y = np.array([[0], [1], [1], [1]])


# Компилируем модель
model.compile(optimizer, loss='binary_crossentropy', metrics=['accuracy'])



# Обучаем модель
model.fit(X, Y, epochs=100)

# Проверяем предсказания
predictions = model.predict(X)
print(predictions)
```

2. Обучить ИНС для вычисления импликации

```python
# Генерируем обучающие данные для импликации
X_train_impl = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_train_impl = np.array([[1], [1], [0], [1]])  # Импликация A -> B

# Создаем модель для импликации
model_impl = Sequential([
    Dense(32, input_shape=(2,)),
    Activation('relu'),
    Dense(1),
    Activation('sigmoid')
])

# Компилируем модель для импликации
model_impl.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Обучаем модель для импликации
model_impl.fit(X_train_impl, y_train_impl, epochs=100)

# Проверяем предсказания для импликации
predictions_impl = model_impl.predict(X_train_impl)
print(predictions_impl)
```

3. Обучить ИНС для вычисления сложения по модулю два

```python
# Генерируем обучающие данные для сложения по модулю два
X_train_add = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_train_add = np.array([[0], [1], [1], [0]])  # Сложение по модулю два

# Создаем модель для сложения по модулю два
model_add = Sequential([
    Dense(32, input_shape=(2,)),
    Activation('relu'),
    Dense(1),
    Activation('sigmoid')
])

# Компилируем модель для сложения по модулю два
model_add.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Обучаем модель для сложения по модулю два
model_add.fit(X_train_add, y_train_add, epochs=100)

# Проверяем предсказания для сложения по модулю два
predictions_add = model_add.predict(X_train_add)
print(predictions_add)
```
