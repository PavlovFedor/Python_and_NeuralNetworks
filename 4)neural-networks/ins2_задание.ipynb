# Подбор параметров для вычисления логических функций средствами однослойных искусственных нейронных сетей

В качестве примера построим и обучим искусственную нейронную сеть (ИНС) вычислению конъюнкции.

Сначала подгрузим библиотеки с линейной моделью слоев Sequential, со слоями Dense

```python
from keras.models import Sequential
from keras.layers import Dense, Activation
```

Создадим модель ИНС и добавим в нее один слой с одним нейроном (units = 1) и двумя входами (input_dim = 2)

```python
from keras import activations
model = Sequential()
model.add(Dense(units = 1, input_dim = 2, activation='sigmoid'))
```

Скомпилируем, получившуюся ИНС

```python
model.compile(optimizer='sgd', loss='mse', metrics=['binary_accuracy'])
```

Создадим обучающий набор X и вектор значений целевой функции (конъюнкции)

```python
import numpy as np
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
X
```

```python
y = np.array([0, 0, 0, 1])
y
```

Обучим ИНС

```python
model.fit(X, y, epochs=5)
```

Протестируем ИНС

```python
model.predict(X)
```

Посмотрим как влияет выбор функции активации на результат классификации

```python
f_act = ['elu', 'softmax', 'selu', 'softplus', 'softsign', 'relu', 'tanh', 'sigmoid', 'hard_sigmoid', 'exponential', 'linear']
for f in f_act:
  model = Sequential()
  model.add(Dense(units = 1, input_dim = 2, activation=f))
  model.compile(optimizer='sgd', loss='mse', metrics=['binary_accuracy'])
  model.fit(X, y, epochs=10)
  yp = model.predict(X)
  print(f, yp)
```

1. Проверьте влияние на результат еще и таких параметров, как optimizer, loss, metrics

```python
optimizers = ['sgd', 'rmsprop', 'adam', 'adagrad']
losses = ['mean_squared_error', 'binary_crossentropy']
metrics = ['accuracy', 'binary_accuracy']

for optimizer in optimizers:
    for loss in losses:
        for metric in metrics:
            model = Sequential()
            model.add(Dense(units=1, input_dim=2, activation='sigmoid'))
            model.compile(optimizer=optimizer, loss=loss, metrics=[metric])
            model.fit(X, y, epochs=10, verbose=0)
            score = model.evaluate(X, y)
            print("Optimizer:", optimizer, "- Loss:", loss, "- Metric:", metric)
            print("Score:", score)
```

2. Скомбинируйте изменение нескольких параметров: activation, optimizer, loss, metrics, -- для проверки их комплексного влияния на результат

```python
for f in f_act:
    for optimizer in optimizers:
        for loss in losses:
            for metric in metrics:
                model = Sequential()
                model.add(Dense(units=1, input_dim=2, activation=f))
                model.compile(optimizer=optimizer, loss=loss, metrics=[metric])
                model.fit(X, y, epochs=10, verbose=0)
                score = model.evaluate(X, y)
                print("Activation:", f, "- Optimizer:", optimizer, "- Loss:", loss, "- Metric:", metric)
                print("Score:", score)
```
