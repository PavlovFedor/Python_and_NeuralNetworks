# Вычисление логических функций средствами однослойных искусственных нейронных сетей

В качестве примера построим и обучим искусственную нейронную сеть (ИНС) вычислению конъюнкции.

Сначала подгрузим библиотеки с линейной моделью слоев Sequential, со слоями Dense

```python
from keras.models import Sequential
from keras.layers import Dense, Activation
```

Создадим модель ИНС и добавим в нее один слой с одним нейроном (units = 1) и двумя входами (input_dim = 2)

```python
from keras import activations
model = Sequential()
model.add(Dense(units = 1, input_dim = 2, activation='sigmoid'))
```

Скомпилируем, получившуюся ИНС

```python
model.compile(optimizer='sgd', loss='mse', metrics=['binary_accuracy'])
```

Создадим обучающий набор X и вектор значений целевой функции (конъюнкции)

```python
import numpy as np
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
X
```

```python
y = np.array([0, 0, 0, 1])
y
```

Обучим ИНС

```python
model.fit(X, y, epochs=5)
```

Протестируем ИНС

```python
model.predict(X)
```
