```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
```

```python

data = {
    'Вакционировано': [10, 490, 500],
    'Не вакционировано': [990, 510,1500],
    'Всего': [1000, 1000, 2000]
}

df = pd.DataFrame(data, index = ['Заболело', 'Не заболело', 'Всего']).T

df
```

```python
# Вычисляем коэффициенты контингенции и ассоциации
chi2, p, _, _ = chi2_contingency(df.values)
cramers_v = np.sqrt(chi2 / (df.sum().min() * (min(df.shape) - 1)))

# Вычисляем коэффициенты взаимной сопряженности Чупрова и Пирсона
chuprov = np.sqrt(chi2 / df['Всего'].sum())
pearson = cramers_v
# Вычисляем коэффициент ассоциации (phi-квадрат)
phi_squared = (chi2 / df['Всего'].sum()) / min(df.shape) 

# Выводим коэффициент ассоциации
print(f"Коэффициент ассоциации (phi-квадрат): {phi_squared}")

# Выводим результаты
print(f"Коэффициент контингенции: {cramers_v}")
print(f"Коэффициент взаимной сопряженности Чупрова: {chuprov}")
print(f"Коэффициент корреляции Пирсона: {pearson}")

print()

# Анализ силы корреляционной зависимости
if chuprov >= 0.8:
    print("Коэффициент взаимной сопряженности Чупрова находится на очень высоком уровне по шкале Чеддока.")
elif chuprov >= 0.6:
    print("Коэффициент взаимной сопряженности Чупрова находится на уровне выше среднего по шкале Чеддока.")
elif chuprov >= 0.4:
    print("Коэффициент взаимной сопряженности Чупрова находится на среднем уровне по шкале Чеддока.")
elif chuprov >= 0.2:
    print("Коэффициент взаимной сопряженности Чупрова находится на уровне ниже среднего по шкале Чеддока.")
else:
    print("Коэффициент взаимной сопряженности Чупрова находится на очень низком уровне по шкале Чеддока.")





```
