# Fourier Series and Square Wave Approximation

## Теоретическое введение

Ряды Фурье - это один из самых интригующих математических инструментов, позволяющий представить периодическую функцию в виде комбинации синусов и косинусов.

### Условия для разложения в ряд Фурье

Для приближения функции рядом Фурье необходимо:

1. Периодическая ограниченная функция с периодом $T$
2. Функция должна быть интегрируемой на периоде

### Формула ряда Фурье

Если условия выполнены, функцию можно представить как:

$$f(x) = {a_0} + \sum_{n=1}^{\infty} \left[ a_n \cos\left(\frac{2\pi n x}{T}\right) + b_n \sin\left(\frac{2\pi n x}{T}\right) \right]$$

**Пояснение**: Это бесконечная сумма, где большее количество членов дает лучшее приближение.

### Коэффициенты Фурье

Коэффициенты вычисляются по формулам:

$$a_0 = \int_{0}^{T} f(x)  dx$$

**Пояснение**: Постоянная составляющая функции

$$a_n = \frac{2}{T} \int_{-T/2}^{T/2} f(x) \cos\left(\frac{2\pi n x}{T}\right)  dx$$

**Пояснение**: Коэффициенты для косинусных составляющих

$$b_n = \frac{2}{T} \int_{-T/2}^{T/2} f(x) \sin\left(\frac{2\pi n x}{T}\right)  dx$$

**Пояснение**: Коэффициенты для синусных составляющих

### Особенности прямоугольного сигнала

Для прямоугольной волны (square wave) коэффициенты упрощаются:

$$a_0 = 0$$
$$a_n = 0 \quad \text{для всех } n$$
$$b_n = \begin{cases} 
\frac{4}{\pi n} & \text{если } n \text{ нечетное} \\
0 & \text{если } n \text{ четное}
\end{cases}$$

**Пояснение**: Прямоугольный сигнал содержит только нечетные гармоники синуса

## Практическая реализация на Python

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")
```

### Определение прямоугольного сигнала

Прямоугольный сигнал определяется как:

$$f(x) = \begin{cases} 
-1 & \text{для } -T/2 \leq x \leq 0 \\
1 & \text{для } 0 \leq x \leq T/2
\end{cases}$$

с периодичностью $T$

```python
# Параметры
x_ = np.linspace(-20, 20, 10000)
T = 8
armonics = 10

def squareWave(x):
    """Генерация прямоугольного сигнала"""
    global T
    lowerBoundLeft = (-T/2)
    lowerBoundRight = 0
    upperBoundLeft = 0
    upperBoundRight = (T/2)
    one = 1
    negativeOne = -1
    
    while True:
        if (x >= lowerBoundLeft) and (x <= lowerBoundRight):
            return negativeOne
        elif (x >= upperBoundLeft) and (x <= upperBoundRight):
            return one
        else:
            lowerBoundLeft -= T/2
            lowerBoundRight -= T/2
            upperBoundLeft += T/2
            upperBoundRight += T/2
            if one == 1:
                one = -1
                negativeOne = 1
            else:
                one = 1
                negativeOne = -1
```

### Вычисление коэффициентов $b_n$

$$b_n = \frac{4}{\pi n} \quad \text{для нечетных } n$$
$$b_n = 0 \quad \text{для четных } n$$

```python
def bn(n):
    """Вычисление коэффициентов b_n"""
    n = int(n)
    if (n % 2 != 0):  # только нечетные гармоники
        return 4 / (np.pi * n)
    else:
        return 0
```

### Вычисление угловой частоты $\omega_n$

$$\omega_n = \frac{2\pi n}{T}$$

**Пояснение**: Угловая частота n-ой гармоники

```python
def wn(n):
    """Вычисление угловой частоты"""
    global T
    wn = (2 * np.pi * n) / T
    return wn
```

### Реализация ряда Фурье

$$f(x) \approx \sum_{n=1}^{N} b_n \sin(\omega_n x)$$

**Пояснение**: Для прямоугольного сигнала $a_0 = 0$ и $a_n = 0$, поэтому остается только сумма синусов

```python
def fourierSeries(n_max, x):
    """Вычисление частичной суммы ряда Фурье"""
    a0 = 0  # для прямоугольного сигнала
    partialSums = a0
    
    for n in range(1, n_max):
        try:
            partialSums = partialSums + bn(n) * np.sin(wn(n) * x)
        except:
            print("pass")
            pass
    
    return partialSums
```

### Визуализация результатов

```python
# Генерация данных
y = []
f = []

for i in x_:
    y.append(squareWave(i))
    f.append(fourierSeries(armonics, i))

# Построение графиков
plt.figure(figsize=(12, 6))
plt.plot(x_, y, color="blue", label="Исходный сигнал")
plt.plot(x_, f, color="red", label="Аппроксимация Фурье ({} гармоник)".format(armonics))
plt.title("Аппроксимация прямоугольного сигнала рядом Фурье")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

### Анализ результатов

С помощью всего 10 гармоник мы получаем хорошее приближение прямоугольного сигнала. Явление Гиббса наблюдается на скачках функции.

## Аппроксимация пилообразного сигнала

### Теория пилообразного сигнала

Пилообразный сигнал определяется как:

$$f(x) = \frac{2}{T} \left( x - T \left\lfloor\frac{x}{T} + \frac{1}{2}\right\rfloor \right)$$

**Пояснение**: Линейно возрастающая функция с периодическими сбросами

### Ряд Фурье для пилообразного сигнала

Для пилообразного сигнала коэффициенты Фурье:

$$a_0 = 0$$
$$a_n = 0 \quad \text{для всех } n$$
$$b_n = (-1)^{n+1} \frac{2}{\pi n}$$

**Пояснение**: Пилообразный сигнал содержит все гармоники синуса с чередующимися знаками

```python
def sawtoothWave(x):
    """Генерация пилообразного сигнала"""
    global T
    # Нормализация x к периоду [-T/2, T/2]
    x_norm = x % T
    if x_norm > T/2:
        x_norm -= T
    return (2 * x_norm) / T
```

### Коэффициенты Фурье для пилообразного сигнала

$$b_n = (-1)^{n+1} \frac{2}{\pi n}$$

**Пояснение**: Знак чередуется для каждой гармоники

```python
def bn_sawtooth(n):
    """Вычисление коэффициентов b_n для пилообразного сигнала"""
    n = int(n)
    return ((-1) ** (n + 1)) * (2 / (np.pi * n))
```

### Ряд Фурье для пилообразного сигнала

$$f(x) \approx \sum_{n=1}^{N} b_n \sin(\omega_n x)$$

**Пояснение**: Аналогично прямоугольному сигналу, но с другими коэффициентами

```python
def fourierSeriesSawtooth(n_max, x):
    """Вычисление частичной суммы ряда Фурье для пилообразного сигнала"""
    partialSums = 0
    
    for n in range(1, n_max):
        try:
            partialSums = partialSums + bn_sawtooth(n) * np.sin(wn(n) * x)
        except:
            pass
    
    return partialSums
```

```python
# Генерация данных для пилообразного сигнала
y_sawtooth = []
f_sawtooth = []

for i in x_:
    y_sawtooth.append(sawtoothWave(i))
    f_sawtooth.append(fourierSeriesSawtooth(armonics, i))

# Построение графиков
plt.figure(figsize=(12, 6))
plt.plot(x_, y_sawtooth, color="green", label="Пилообразный сигнал")
plt.plot(x_, f_sawtooth, color="orange", label="Аппроксимация Фурье ({} гармоник)".format(armonics))
plt.title("Аппроксимация пилообразного сигнала рядом Фурье")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

### Сравнение обоих сигналов

**Наблюдения**:
- Пилообразный сигнал требует больше гармоник для хорошей аппроксимации
- Явление Гиббса более выражено из-за более сложной формы сигнала
- Чередование знаков коэффициентов обеспечивает лучшую сходимость

```python
# Сравнительный анализ с разным количеством гармоник
harmonics_list = [5, 10, 20, 50]

plt.figure(figsize=(15, 10))

for i, harmonics_count in enumerate(harmonics_list, 1):
    f_sawtooth_temp = []
    for x_val in x_:
        f_sawtooth_temp.append(fourierSeriesSawtooth(harmonics_count, x_val))
    
    plt.subplot(2, 2, i)
    plt.plot(x_, y_sawtooth, color="green", label="Исходный сигнал")
    plt.plot(x_, f_sawtooth_temp, color="orange", label="{} гармоник".format(harmonics_count))
    plt.title("Аппроксимация с {} гармониками".format(harmonics_count))
    plt.legend()
    plt.grid(True)

plt.tight_layout()
plt.show()
```

### Выводы

1. **Прямоугольный сигнал**: Быстрая сходимость, достаточно 10-20 гармоник
2. **Пилообразный сигнал**: Медленная сходимость, требуется 50+ гармоник
3. **Точность аппроксимации**: Зависит от гладкости исходной функции
