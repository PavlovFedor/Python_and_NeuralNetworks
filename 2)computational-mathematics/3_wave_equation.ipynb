# Уравнение волны и его численная реализация

## Теоретическая часть

### Уравнение волны

Рассмотрим идеальную струну длины $L$ и попытаемся найти уравнение, описывающее колебание струны. Предполагая, что струна закреплена на концах и начинает движение из известного положения $f(x)$, самое простое предположение состоит в том, что ускорение каждого участка струны пропорционально кривизне струны:

$$\frac{\partial^2 u}{\partial t^2} = c^2 \frac{\partial^2 u}{\partial x^2}$$

где $u(x,t)$ - смещение струны в точке $x$ в момент времени $t$, а $c$ - скорость волны.

### Общее решение

Уравнение выше является уравнением в частных производных (УЧП), называемым **волновым уравнением**. Оно может использоваться для моделирования различных явлений, таких как вибрация струн и распространение волн.

Общее решение волнового уравнения имеет вид:

$$u(x,t) = f(x - ct) + g(x + ct)$$

где $f$ и $g$ - произвольные функции, определяемые начальными условиями.

### Физическая интерпретация

Решение $f(x - ct)$ представляет волну, движущуюся вправо со скоростью $c$, а $g(x + ct)$ - волну, движущуюся влево со скоростью $c$. Общее решение является суперпозицией этих двух волн.

## Практическая реализация

### Конкретное решение

Рассмотрим следующее конкретное решение:

$$u(x,t) = \frac{1}{2}[\sin(x + ct) + \sin(x - ct)]$$

Это решение удовлетворяет волновому уравнению и представляет суперпозицию двух синусоидальных волн, движущихся в противоположных направлениях.

### Реализация в Python

Теперь реализуем это решение численно с помощью Python и создадим анимацию колебаний струны.

```python
import numpy as np
from numpy import pi
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Используем темный фон для графиков
plt.style.use('dark_background')

# Скорость волны (м/с)
c = 1

# Пространственная ось x от -π до π
x0 = np.linspace(-pi, pi, 1000)  # Уменьшим количество точек для производительности

# Начальное время
t0 = 0

# Шаг по времени
dt = 0.05

# Решение волнового уравнения: u(x,t) = 0.5[sin(x+ct) + sin(x-ct)]
def u(x, t):
    return 0.5 * (np.sin(x + c * t) + np.sin(x - c * t))

# Формула в LaTeX: $$u(x,t) = \frac{1}{2}[\sin(x + ct) + \sin(x - ct)]$$
# Это решение представляет суперпозицию двух синусоидальных волн,
# движущихся в противоположных направлениях со скоростью c

# Создаем массив значений для анимации
frames = []
for i in range(100):  # Уменьшим количество кадров
    value = u(x0, t0)
    t0 = t0 + dt
    frames.append(value)

# Создаем фигуру и оси
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(-pi, pi)
ax.set_ylim(-2, 2)
ax.grid(True)
ax.set_xlabel('Положение x')
ax.set_ylabel('Смещение u(x,t)')
ax.set_title('Волновое уравнение')

# Создаем линию для анимации
line, = ax.plot(x0, frames[0], color='cyan')

# Функция анимации
def animate(i):
    line.set_ydata(frames[i])
    ax.set_title(f'Волновое уравнение: t = {i*dt:.2f} с')
    return line,

# Создаем анимацию
anim = animation.FuncAnimation(fig, animate, frames=len(frames), interval=50, blit=True)

# Отображаем анимацию в Jupyter/Colab
plt.close(fig)  # Чтобы избежать дублирования отображения
HTML(anim.to_jshtml())
```

### Объяснение кода

1. **Импорт библиотек**: Используем numpy для численных вычислений, matplotlib для визуализации, и IPython.display для отображения анимации в ноутбуке.

2. **Параметры модели**: 
   - $c = 1$ м/с - скорость волны
   - $x \in [-\pi, \pi]$ - пространственная область
   - $\Delta t = 0.05$ с - шаг по времени

3. **Решение уравнения**: 
   $$u(x,t) = \frac{1}{2}[\sin(x + ct) + \sin(x - ct)]$$
   
4. **Анимация**: Визуализируем распространение волны во времени, используя эффективный метод с `blit=True` для лучшей производительности.
