# Уравнение переноса с затуханием
## Теоретическая часть

### Уравнение переноса

Уравнение переноса описывает распространение волны с постоянной скоростью $k$:

$$\frac{\partial u}{\partial t} + k \frac{\partial u}{\partial x} = 0$$

Общее решение имеет вид:

$$u(x,t) = f(x - kt)$$

где $f$ - произвольная функция.

Для гауссовой функции, решение принимает вид:

$$u(x,t) = e^{-(x-kt)^2}$$

### Уравнение переноса с затуханием

Чтобы учесть затухание (диссипацию энергии), добавляем член с затуханием:

$$\frac{\partial u}{\partial t} + k \frac{\partial u}{\partial x} = -\gamma u$$

где $\gamma > 0$ - коэффициент затухания. Общее решение:

$$u(x,t) = f(x - ct) \cdot e^{-\gamma t}$$

Решение представляет собой бегущую волну, которая экспоненциально затухает со временем.

## Практическая реализация
### Импорт библиотек и настройка

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
import scipy.integrate as integrate

# Настройка графики
%matplotlib inline
plt.rcParams['figure.figsize'] = [10, 6]
plt.rcParams['font.size'] = 12
```

### Параметры модели и начальные условия

Уравнение с затуханием: $$u(x,t) = e^{-(x-kt)^2} \cdot e^{-\gamma t}$$

```python
# Начальные условия
x0 = np.linspace(-2, 15, 10000)  # Пространственная сетка
t0 = 0  # Начальное время

# Параметры
dt = 0.01  # Шаг по времени
k = 3  # Скорость переноса (k > 0)
gamma = 2  # Коэффициент затухания (γ > 0)
A = 4  # Амплитуда

# Функция решения уравнения переноса с затуханием
def u(x, t):
    """
    Решение уравнения переноса с затуханием:
    u(x,t) = A * exp(-(x-kt)²) * exp(-γt)
    """
    return A * np.exp(-(x - k * t)**2) * np.exp(-gamma * t)
```

### Подготовка данных для анимации

```python
# Подготовка данных для анимации
t_values = []  # Список значений времени
u_values = []  # Список значений функции u(x,t)

current_time = t0
for i in range(500):
    u_values.append(u(x0, current_time))  # Вычисление u(x,t) для текущего времени
    t_values.append(current_time)  # Сохранение текущего времени
    current_time += dt  # Увеличение времени на шаг dt
```

### Создание анимации (работает в Google Colab)

```python
# Создание фигуры и оси
fig, ax = plt.figure(), plt.axes()
fig.set_dpi(100)

# Настройка графика
ax.set_xlim(-2, 15)
ax.set_ylim(-1, A + 1)
ax.grid(True)
ax.set_xlabel('x')
ax.set_ylabel('u(x,t)')
ax.set_title('Уравнение переноса с затуханием')

# Создание линии для анимации
line, = ax.plot([], [], color='red')
time_text = ax.text(0.02, 0.95, '', transform=ax.transAxes)

def init():
    """Инициализация анимации"""
    line.set_data([], [])
    time_text.set_text('')
    return line, time_text

def animate_frame(i):
    """Обновление кадра анимации"""
    if i < len(u_values):
        line.set_data(x0, u_values[i])
        time_text.set_text(f'Время: {t_values[i]:.2f} с')
    return line, time_text

# Создание анимации
anim = animation.FuncAnimation(
    fig, animate_frame, init_func=init,
    frames=min(360, len(u_values)), interval=20, blit=True
)

plt.close(fig)

# Отображение анимации в Colab
HTML(anim.to_jshtml())
```

### Интерактивное исследование параметров

Исследуем, как изменение параметров влияет на решение:

```python
# Исследование влияния коэффициента затухания
gamma_values = [0.5, 1, 2, 4]  # Разные значения γ
t_study = 1.0  # Фиксированное время для исследования

plt.figure(figsize=(10, 6))
for gamma_val in gamma_values:
    def u_study(x, t):
        return A * np.exp(-(x - k * t)**2) * np.exp(-gamma_val * t)
    
    plt.plot(x0, u_study(x0, t_study), label=f'γ = {gamma_val}')

plt.grid(True)
plt.legend()
plt.title(f'Влияние коэффициента затухания при t={t_study}')
plt.xlabel('x')
plt.ylabel('u(x,t)')
plt.xlim([-2, 15])
plt.show()
```
