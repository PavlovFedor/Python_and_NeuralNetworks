# Численное интегрирование с Python

## Теоретическое объяснение

Численное интегрирование - это семейство алгоритмов для вычисления численного значения определенного интеграла. Два простейших метода для выполнения численного интегрирования - это метод прямоугольников и метод трапеций.

### Определенный интеграл

Для функции $f(x) = x^2$ определенный интеграл от $a$ до $b$ вычисляется по формуле:

$$\int_a^b x^2 dx = \frac{b^3 - a^3}{3}$$

### Метод прямоугольников

Метод прямоугольников аппроксимирует площадь под кривой суммой площадей прямоугольников:

$$\int_a^b f(x) dx \approx \sum_{i=1}^{n} f\left(\frac{x_{i-1} + x_i}{2}\right) \cdot (x_i - x_{i-1})$$

где $n$ - количество точек разбиения.

### Метод трапеций

Метод трапеций аппроксимирует площадь под кривой суммой площадей трапеций:

$$\int_a^b f(x) dx \approx \sum_{i=1}^{n} \frac{f(x_{i-1}) + f(x_i)}{2} \cdot (x_i - x_{i-1})$$

### Погрешность методов

Относительная погрешность вычисляется как:

$$\text{Погрешность} = \frac{\text{Приближенное значение} - \text{Точное значение}}{\text{Точное значение}} \times 100\%$$

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')

# Исходная функция f(x) = x²
def f(x):
    return x**2

# Точное значение определенного интеграла от a до b
# ∫ₐᵇ x² dx = (b³ - a³)/3
def F(a, b):
    if a > b:
        raise ValueError('b must be greater than a')
    elif a == b:
        return 0
    else:
        return (b**3 - a**3) / 3
```

## Реализация методов численного интегрирования

### Метод прямоугольников

Формула метода прямоугольников:

$$\text{miniArea}(c,d) = (d-c) \cdot f\left(\frac{c+d}{2}\right)$$

### Метод трапеций

Формула метода трапеций:

$$\text{miniArea}(c,d) = (d-c) \cdot \frac{f(c) + f(d)}{2}$$

```python
def approximateNumerical(a, b, points=10, error=False, mod='rectangular', plt_data=False):
    if points < 2:
        raise ValueError('Number of points must be greater than 2')
    if a == b:
        return 0
    
    # Создаем равномерно распределенные точки на интервале [a, b]
    n = np.linspace(a, b, points)
    partialSum = 0
    
    # Выбираем метод интегрирования
    if mod == 'rectangular':
        def miniArea(c, d):
            return (d - c) * f((c + d) / 2)  # Метод прямоугольников
    elif mod == 'trapezoidal':
        def miniArea(c, d):
            return (d - c) * (f(c) + f(d)) / 2  # Метод трапеций
    else:
        raise ValueError('Method ' + mod + ' unknown')
    
    # Вычисляем сумму площадей
    for i in range(1, len(n)):
        partialSum += miniArea(n[i-1], n[i])
    
    # Вычисляем относительную погрешность в процентах
    e = (partialSum - F(a, b)) / F(a, b) * 100
    
    if error:
        print('\nАппроксимация с использованием метода ' + mod + '...')
        print('Процентная погрешность: ', e, '%')
    
    if plt_data:
        plot_dat(a, b, points, mod=mod)
    
    return partialSum, e

# Функция для визуализации методов интегрирования
def plot_dat(a, b, points, mod='rectangular'):
    n = np.linspace(a, b, points)
    plt.plot(n, f(n), color='red')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Численная аппроксимация: ' + mod)
    
    if mod == 'rectangular':
        for i in range(1, len(n)):
            c = n[i-1]
            d = n[i]
            plt.plot([c, c], [0, f((c + d) / 2)], color='blue')
            plt.plot([d, d], [0, f(d)], color='blue')
            plt.plot([c, d], [f((c + d) / 2), f((c + d) / 2)], color='blue')
        plt.show()
    
    if mod == 'trapezoidal':
        for i in range(1, len(n)):
            c = n[i-1]
            d = n[i]
            plt.plot([d, d], [0, f(d)], color='blue')
            plt.plot([c, c], [0, f(c)], color='blue')
            plt.plot([c, d], [f(c), f(d)], color='blue')
        plt.show()
    
    return 0

# Функция для аппроксимации с заданной точностью
def approxGivenPrecision(a, b, error=0.5, mod='rectangular', printf=False):
    e = 100  # Начальная погрешность
    p = 5    # Начальное количество точек
    itMax = 500  # Максимальное количество итераций
    it = 0       # Счетчик итераций
    
    while abs(e) > error:
        area, e = approximateNumerical(a, b, mod=mod, points=p)
        p += 10  # Увеличиваем количество точек
        it += 1  # Увеличиваем счетчик итераций
        
        if printf:
            print('Аппроксимация с использованием метода ' + mod + ' с ' + str(p) + ' точками, процентная погрешность: ' + str(e) + '%')
        
        if it > itMax:
            print('Превышено максимальное количество итераций: ' + str(itMax))
            break
    
    return area
```

## Запуск программы и результаты

### Параметры вычислений:
- Нижний предел интегрирования: $m = 0$
- Верхний предел интегрирования: $M = 25$
- Количество точек: $p = 50$

### Точное значение интеграла:
$$\int_0^{25} x^2 dx = \frac{25^3 - 0^3}{3} = \frac{15625}{3} \approx 5208.333$$

```python
# Запуск программы с параметрами по умолчанию
m = 0
M = 25
p = 50

print('Точная площадь:', F(m, M))
print('Аппроксимация методом прямоугольников:', approximateNumerical(m, M, p, error=True, plt_data=True))
print('Аппроксимация методом трапеций:', approximateNumerical(m, M, p, mod='trapezoidal', error=True, plt_data=True))

print('#############################################################')

# Аппроксимация с заданной точностью 0.05%
print('Аппроксимация с точностью 0.05% методом прямоугольников:')
print(approxGivenPrecision(m, M, error=0.05, printf=True))

print('Аппроксимация с точностью 0.05% методом трапеций:')
print(approxGivenPrecision(m, M, error=0.05, mod='trapezoidal', printf=True))
```

## Анализ результатов

Как видно из результатов:

1. **Метод прямоугольников** дает небольшую отрицательную погрешность (-0.01%)
2. **Метод трапеций** дает небольшую положительную погрешность (+0.02%)
3. Для достижения точности 0.05% метод прямоугольников требует меньше точек, чем метод трапеций
